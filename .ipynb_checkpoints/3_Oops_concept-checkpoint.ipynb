```python
# Create a function that checks whether the given number is Odd or Even

def OddEven():
    data = int(input("Enter a number:"))
    print('Enter a number:',data)
    if((data % 2) == 0):
        print(data,' is Even number')
        return data
    else:
        print(data,' is Odd number')
        return data
```

```python
OddEven()
```

```python
# Create a function that tells elegibility of marriage for male and female according to their age limit like 21 for male and 18 for female

def marriage():
    gender = input('enter the gender :')
    age = int(input('enter the age :'))
    if(gender == 'Male' and age >= 21):
        print('Your Gender:',gender)
        print('Your Age:',age)
        print("ELIGIBLE")
    elif(gender == 'Female' and age >= 18):
        print('Your Gender:',gender)
        print('Your Age:',age)
        print("ELIGIBLE")
    else:
        print('age limit like 21 for male and 18 for female : ',gender,age)
        print("NOT ELIGIBLE")
        
        
```

```python
marriage()
```

```python
def calculate_payout(salary):
    if salary > 50000:
        bonus = salary * 0.10
    else:
        bonus = 0.0
    total_payout = salary + bonus
    return bonus, total_payout    

employees = [
    {"name": "Alice", "salary": 65000},
    {"name": "Bob", "salary": 48000},
    {"name": "Charlie", "salary": 52000},
    {"name": "Diana", "salary": 35000},
    {"name": "Evan", "salary": 75000},
]
for emp in employees:
    bonus, total = calculate_payout(emp["salary"])
    print(
        f"{emp['name']:<10} | ${emp['salary']:<9,.2f} | ${bonus:<9,.2f} | ${total:<10,.2f}"
    )

```

```python
def check_leap_year(year):
    # Step 2: If the year is divisible by 400, it is a leap year
    if year % 400 == 0:
        return True
    # Step 3: If the year is divisible by 100 (and not 400), it is not a leap year
    elif year % 100 == 0:
        return False
    # Step 4: If the year is divisible by 4 (and not 100), it is a leap year
    elif year % 4 == 0:
        return True
    # Step 5: Otherwise, it is not a leap year
    else:
        return False


# Step 1: Read the input year
input_year = int(input("Enter a year: "))

if check_leap_year(input_year):
    print(f"{input_year} is a leap year.")
else:
    print(f"{input_year} is NOT a leap year.")
```
